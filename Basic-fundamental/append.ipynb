## checking memory address

```python
l1 = ["Baa", "Baa", "Black", "Sheep"]
print(id(l1))
```

```python
l1.append("Baa")
print(id(l1))
```

## set

```python
s1 = {"Baa", "Baa", "Black", "Sheep"}
print(s1)
print(id(s1))

s1.add("Python")
print(s1)
print(id(s1))
```

## concate

```python
# List concatenation creates a new list; l1 remains unchanged
list2 = l1 + ["Python"]
print("Original list:", l1)
print("New list:     ", list2)
print("IDs:", id(l1), id(list2))  # Different IDs mean these are separate lists

# Tuple concatenation
tuple2 = t1 + ("Python",)
print(tuple2)

# Set union
set2 = s1 | {"Notebook", "Python"}
print(set2)

# Dictionary merge (Python 3.9+)
dict1 = {"language": "Python", "version": 3}
dict2 = {"editor": "Jupyter", "version": 4}
dict_merged = dict1 | dict2  # dict2's value wins for duplicate keys
print(dict_merged)
```
